```python
import matplotlib.pyplot as plt
import networkx as nx

# Create a directed graph
G = nx.DiGraph()

# Define nodes
nodes = [
    "Income/Spending_Decile", "Location", "Device_Type", "Festival_Period",
    "Brand_Promotions", "Cashback_Offer", "Transactions", "Merchant_Diversity",
    "Product_Category", "Payment_Diversity", "Spending_Concentration_HHI"
]
G.add_nodes_from(nodes)

# Define edges (causal paths)
edges = [
    ("Income/Spending_Decile", "Cashback_Offer"),
    ("Income/Spending_Decile", "Transactions"),
    ("Location", "Cashback_Offer"),
    ("Location", "Transactions"),
    ("Device_Type", "Cashback_Offer"),
    ("Device_Type", "Transactions"),
    ("Festival_Period", "Cashback_Offer"),
    ("Festival_Period", "Transactions"),
    ("Brand_Promotions", "Cashback_Offer"),
    ("Cashback_Offer", "Transactions"),
    ("Merchant_Diversity", "Transactions"),
    ("Product_Category", "Transactions"),
    ("Payment_Diversity", "Transactions"),
    ("Spending_Concentration_HHI", "Transactions")
]
G.add_edges_from(edges)

# Set layout
pos = nx.spring_layout(G, k=1.5, seed=42)

# Draw the DAG
plt.figure(figsize=(12, 8))
nx.draw(
    G, pos, with_labels=True, node_size=3000, node_color="lightblue",
    font_size=9, font_weight="bold", edgecolors="black", arrowsize=20
)

plt.title("Causal DAG for Cashback Effect on UPI Transactions", fontsize=14, fontweight="bold")
plt.show()
```

```python
import pandas as pd
import statsmodels.api as sm

# Load dataset
df = pd.read_csv("dataset_ada.csv")  # replace with your path

# Aggregate per user
agg_df = df.groupby('user_id').agg(
    Transactions=('transaction_id', 'count'),
    Cashback_Offer=('cashback', lambda x: 1 if x.sum() > 0 else 0),  # binary treatment
    Product_Amount=('product_amount', 'sum'),
    Transaction_Fee=('transaction_fee', 'sum'),
    Loyalty_Points=('loyalty_points', 'sum'),
    Location=('location', 'first'),
    Device_Type=('device_type', 'first'),
    Transaction_Status=('transaction_status', 'first'),
    Payment_Method=('payment_method', 'first')
).reset_index()

# Diversity metrics
agg_df['Merchant_Diversity'] = df.groupby('user_id')['merchant_id'].nunique().values
agg_df['Payment_Diversity'] = df.groupby('user_id')['payment_method'].nunique().values

# Select predictors and outcome
X = agg_df[['Cashback_Offer', 'Location', 'Device_Type', 'Merchant_Diversity',
            'Product_Amount', 'Payment_Diversity', 'Transaction_Fee',
            'Loyalty_Points', 'Transaction_Status', 'Payment_Method']]
y = agg_df['Transactions']

# Step 1: One-hot encode all categorical columns
categorical_cols = X.select_dtypes(include=['object']).columns.tolist()
X = pd.get_dummies(X, columns=categorical_cols, drop_first=True)

# Step 2: Convert all boolean columns to int
bool_cols = X.select_dtypes(include=['bool']).columns.tolist()
X[bool_cols] = X[bool_cols].astype(int)

# Step 3: Ensure all numeric
X = X.apply(pd.to_numeric)
y = pd.to_numeric(y, errors='coerce').fillna(0)

# Step 4: Add intercept
X = sm.add_constant(X)

# Step 5: Fit OLS
model = sm.OLS(y, X).fit()

# Step 6: Output results
print(model.summary())
print(f"\nEstimated effect of cashback on transactions: {model.params['Cashback_Offer']:.2f}")
```

```python
# -----------------------------
# 1. Import libraries
# -----------------------------
import pandas as pd
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

# -----------------------------
# 2. Load your dataset
# -----------------------------
# Replace this with your actual data loading
# df = pd.read_csv('your_dataset.csv')

# Outcome variable
y = df['Transactions']  # numeric, e.g., transaction count or volume

# -----------------------------
# 3. Select predictors (treatment + confounders)
# -----------------------------
predictors = ['cashback', 'Merchant_Diversity', 'Product_Amount', 'Payment_Diversity',
              'transaction_fee', 'loyalty_points', 'location', 'device_type',
              'transaction_status', 'payment_method']

X = df[predictors]

# -----------------------------
# 4. One-hot encode categorical variables
# -----------------------------
categorical_cols = ['location', 'device_type', 'transaction_status', 'payment_method']
X = pd.get_dummies(X, columns=categorical_cols, drop_first=True)  # avoids dummy variable trap

# -----------------------------
# 5. Convert all predictors to numeric
# -----------------------------
X = X.astype(float)

# -----------------------------
# 6. Check multicollinearity (VIF)
# -----------------------------
vif_data = pd.DataFrame()
vif_data['feature'] = X.columns
vif_data['VIF'] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
print("VIF values:\n", vif_data)
# Optional: if VIF > 10, consider combining categories or noting multicollinearity in report

# -----------------------------
# 7. Add intercept
# -----------------------------
X = sm.add_constant(X)

# -----------------------------
# 8. Fit OLS regression
# -----------------------------
model = sm.OLS(y, X).fit()

# -----------------------------
# 9. Results
# -----------------------------
print(model.summary())

# Estimated effect of cashback on transactions
print(f"\nEstimated effect of cashback on transactions: {model.params['cashback']:.4f}")
```
